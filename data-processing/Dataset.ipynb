```python
inventory_df['product_name'] = inventory_unique_data_df['product_name'].values
inventory_df['category'] = inventory_unique_data_df['category'].values
inventory_df['unit_price'] = inventory_unique_data_df['unit_price'].values
```

```python
#Because we changed unit_price we know need to make sure unit_cost is below that. We will give random profit margin of 25-40 with the average being closer to 35%
inventory_df['unit_cost'] = (inventory_df['unit_price'] / (1 + np.random.uniform(0.25, 0.40, size=len(inventory_df)))).round(2)
```
